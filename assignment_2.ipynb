```python
#body temperature as the threshhold
bodytemp_fahr=98.6
cels=float(input("Enter temperature in degree Celcius:"))
#converting temperature in degree celsius to Fahrenheit
fahr=(cels*1.8)+32
print(fahr)
if fahr < bodytemp_fahr:
  print("Body temperature is below threshold")
elif fahr > bodytemp_fahr:
  print("Body temperature is above threshold")
else:
  print("Body temperature is within threshold")

```

```python
temps_celsius=[23,36,37.5,43]
#iterates the list converting all temps in degree ccelsius to fahrenheit
for temp in temps_celsius:
  #convertion
  fahr=(temp*1.8)+32
  print(fahr)

```

```python
def celsius_to_fahrenheit(celsius):
 #Converts temperature from Celsius to Fahrenheit.
  return (celsius * 1.8) + 32

# Loop through the list of Celsius temperatures
temps=[23,36,37.5,43]
for temp in temps:
  temp_f = celsius_to_fahrenheit(temp)
  print(f"Temperature {temp}°C is {temp_f}°F")

  # Check against the body temperature threshold
  if temp_f < bodytemp_fahr:
    print("  Body temperature is below threshold")
  elif temp_f > bodytemp_fahr:
    print("  Body temperature is above threshold")
  else:
    print("  Body temperature is within threshold")
```
